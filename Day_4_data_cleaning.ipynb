```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
# load the dataset
data=pd.read_csv("portland_housing.csv")
# print the shape of the data
print(data.shape)
# Print the first five rows of the data
print(data.head())
```

```python

# check the sum of null values in the DataclassInstance
print(data.isnull().sum())
```

```python
# check the data types of the dataset collumns
print(data.dtypes)
```

```python
bath_mode = data["bathrooms"].mode()[0]
data["bathrooms"] = data["bathrooms"].fillna(bath_mode)
bed_mean=data["bedrooms"].mean()
data["bedrooms"]=data["bedrooms"].fillna(bed_mean)
year_mode=data["yearBuilt"].mode()[0]
data["yearBuilt"]=data["yearBuilt"].fillna(year_mode)
zestim_mean=data["zestimate"].mean()
data["zestimate"]=data["zestimate"].fillna(zestim_mean)
zestim_h=data["zestimateHighPercent"].mode()[0]
data["zestimateHighPercent"]=data["zestimateHighPercent"].fillna(zestim_h)
zestim_l=data["zestimateLowPercent"].mode()[0]
data["zestimateLowPercent"]=data["zestimateLowPercent"].fillna(zestim_l)
zpid_mean=data["zpid"].mean()
data["zpid"]=data["zpid"].fillna(zpid_mean)



print(data.isnull().sum())
```
